```python

import pandas as pd
import openpyxl
```

```python
# read in workpapers_testing.xlsx
# this code reads in the workpapers which have a unique sheet for every condition 
# it creates a dataframe where each row is one of the unique sheets
# the columns names are what were in the first column of each workpaper sheet
# the values are what was in the second column of each sheet
# only down to row 100 in each excel file


# Load the workbook
workbook_path = './001_input/workpapers_testing.xlsx'
wb = openpyxl.load_workbook(workbook_path)

# List to store data from each sheet
all_sheets_data = []

# Iterate through all sheets
for sheet_name in wb.sheetnames:
    ws = wb[sheet_name]
    
    # Create dictionary for this sheet
    sheet_dict = {'sheet_name': sheet_name}
    
    # Read first 100 rows, columns A (field) and B (value)
    for row in range(2, min(102, ws.max_row + 1)):  # Start at row 2 (skip header), max 100 rows
        field_cell = ws.cell(row=row, column=1)  # Column A
        value_cell = ws.cell(row=row, column=2)  # Column B
        
        if field_cell.value:  # Only add if field name exists
            sheet_dict[field_cell.value] = value_cell.value
    
    all_sheets_data.append(sheet_dict)

# Create DataFrame where each row is a sheet
df_workpapers = pd.DataFrame(all_sheets_data)

# Display shape and first few rows
print(f"DataFrame shape: {df_workpapers.shape}")
df_workpapers.head()
```

```python
# make all column headers snake case and all values snake case
# list all the columns in df_workpapers

# Function to convert to snake_case
def to_snake_case(text):
    if text is None or pd.isna(text):
        return text
    text = str(text)
    # Replace spaces, dots, slashes, and other special characters with underscores
    text = text.lower().replace(' ', '_').replace('$', 'usd').replace('/', 'per').replace('&', 'and').replace('.', '')
    # Remove any duplicate underscores
    while '__' in text:
        text = text.replace('__', '_')
    return text

# Convert column names to snake_case
df_workpapers.columns = [to_snake_case(col) for col in df_workpapers.columns]

# Convert all string values to snake_case
for col in df_workpapers.columns:
    if df_workpapers[col].dtype == 'object':  # Only process string columns
        df_workpapers[col] = df_workpapers[col].apply(to_snake_case)

# List all columns
print(f"Total columns: {len(df_workpapers.columns)}\n")
print("All columns:")
print(df_workpapers.columns.tolist())
```
